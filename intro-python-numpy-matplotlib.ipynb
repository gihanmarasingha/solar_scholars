# 1. Variables, Operators, Assignment

A variable is a name attached to a value.

Python evaluates the right-hand side first, then stores the result in the name on the left.


````python
distance_km = 12
minutes = 48

speed_km_per_hour = distance_km / (minutes / 60)

print(speed_km_per_hour)
````

# 2. Function Definitions

A function packages a reusable calculation.

Core syntax:

```python
def function_name(input_1, input_2):
    result = ...
    return result
```


````python
def convert_celsius_to_kelvin(celsius):
    kelvin = celsius + 273.15
    return kelvin

print(convert_celsius_to_kelvin(20))
````

## Functions Make Intent Clear

Use names that describe the calculation, not just the implementation.


````python
def circle_area(radius):
    return 3.14159 * radius ** 2

area_small = circle_area(2)
area_large = circle_area(10)

print(area_small)
print(area_large)
````

## Function Arguments

Arguments are local names inside the function.

Changing a variable name outside the function does not change the function definition.


````python
def linear_model(x, slope, intercept):
    return slope * x + intercept

print(linear_model(5, slope=2, intercept=1))
````

# 5. Making Plots with Matplotlib

Typical pattern:

1. create `x` and `y`
2. call a plotting function
3. label the plot
4. show it


````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.title("A sine wave")
plt.show()
````

## Scatter Plots

Use scatter plots for observed data points.


````python
x_data = np.array([0, 1, 2, 3, 4, 5])
y_data = np.array([1.1, 2.0, 2.9, 4.2, 5.1, 5.8])

plt.scatter(x_data, y_data)
plt.xlabel("x")
plt.ylabel("observed y")
plt.title("Measured data")
plt.show()
````

## Plot Several Lines

Use labels and `plt.legend()` when showing more than one series.


````python
x = np.linspace(0, 5, 100)

plt.plot(x, x, label="linear")
plt.plot(x, x ** 2, label="quadratic")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
````

# 6. Iteration and Conditionals

Programs become useful when they can repeat work and make decisions.


## `for` Loops

A `for` loop repeats the same block for each item in a sequence.


````python
for value in [1, 2, 3, 4]:
    squared = value ** 2
    print(value, squared)
````

## Building a List in a Loop

Start with an empty list, then append results.


````python
squares = []

for value in range(5):
    squares.append(value ** 2)

print(squares)
````

## `if`, `elif`, `else`

Conditionals choose which block of code runs.


````python
temperature = 18

if temperature < 10:
    label = "cold"
elif temperature < 20:
    label = "mild"
else:
    label = "warm"

print(label)
````

## Combining Loops and Conditions


````python
numbers = np.arange(10)

even_squares = []
for number in numbers:
    if number % 2 == 0:
        even_squares.append(number ** 2)

print(even_squares)
````

# 7. Syntax of `curve_fit`

`curve_fit` estimates model parameters from data.

It answers: which parameter values make this function best match these observations?


## The Required Pieces

1. A model function: `model(x, parameter_1, parameter_2, ...)`
2. Observed `x` values
3. Observed `y` values
4. A call to `curve_fit(model, x_data, y_data)`


````python
from scipy.optimize import curve_fit

# Model function: x must be the first argument.
def straight_line(x, slope, intercept):
    return slope * x + intercept

x_data = np.array([0, 1, 2, 3, 4, 5])
y_data = np.array([1.1, 2.0, 2.9, 4.2, 5.1, 5.8])

parameters, covariance = curve_fit(straight_line, x_data, y_data)

slope, intercept = parameters
print(slope, intercept)
````

## Plot Data and Fitted Curve

Fit parameters become inputs to the model function.


````python
x_fit = np.linspace(x_data.min(), x_data.max(), 100)
y_fit = straight_line(x_fit, slope, intercept)

plt.scatter(x_data, y_data, label="data")
plt.plot(x_fit, y_fit, label="fit")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
````

## Nonlinear Example

The syntax is the same for nonlinear models.


````python
def exponential_decay(x, amplitude, rate, offset):
    return amplitude * np.exp(-rate * x) + offset

x_data = np.linspace(0, 5, 12)
y_data = np.array([5.2, 4.1, 3.3, 2.7, 2.3, 1.9, 1.7, 1.5, 1.4, 1.3, 1.2, 1.1])

initial_guess = [4, 0.8, 1]
parameters, covariance = curve_fit(exponential_decay, x_data, y_data, p0=initial_guess)

print(parameters)
````

## Common Curve Fit Pitfall

The model function must take `x` first, then the parameters.

Correct:

```python
def model(x, a, b):
    return a * x + b
```

Incorrect:

```python
def model(a, b, x):
    return a * x + b
```


## Optional Extension

Use `np.random.normal` to add noise to artificial data, then fit the original model back to the noisy data.


````python
rng = np.random.default_rng(42)

x_data = np.linspace(0, 10, 30)
y_true = straight_line(x_data, slope=2, intercept=1)
y_noisy = y_true + rng.normal(loc=0, scale=2, size=len(x_data))

parameters, covariance = curve_fit(straight_line, x_data, y_noisy)
print(parameters)
````

## Finally, let's try fitting a simple curve to data using curvefit


````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit



# Define the True Function
def linear_function(x, m, c):
    """A simple linear function: y = m*x + c"""
    return m * x + c

# Generate Synthetic Data
# True parameters for our linear function
true_m = 2.5
true_c = 5.0

# Generate x values
x_data = np.linspace(0, 10, 50) # 50 points between 0 and 10

# Generate ideal y values
y_ideal = linear_function(x_data, true_m, true_c)

# Add some random noise to the y values
noise = np.random.normal(0, 1.5, len(x_data)) # Mean 0, std dev 1.5
y_data = y_ideal + noise

# Plot the generated data
plt.figure(figsize=(8, 6))
plt.scatter(x_data, y_data, label='Noisy Data', s=20)
plt.plot(x_data, y_ideal, color='red', linestyle='--', label='Ideal True Function')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Generated Data with Noise (Linear)')
plt.legend()
plt.grid(True)

# 4. Perform Curve Fitting with curve_fit
# Perform the curve fit
popt, pcov = curve_fit(linear_function, x_data, y_data)

# popt contains the optimal parameters (m_fit, c_fit)
m_fit, c_fit = popt

print("--- Linear Fit Results ---")
print(f"Fitted slope (m): {m_fit:.3f}")
print(f"Fitted intercept (c): {c_fit:.3f}")
print(f"True slope (m): {true_m}")
print(f"True intercept (c): {true_c}")
print("-" * 25)

# 5. Visualize the Fitted Curve
# Generate y values using the fitted parameters
y_fit = linear_function(x_data, m_fit, c_fit)

plt.figure(figsize=(10, 7))
plt.scatter(x_data, y_data, label='Noisy Data', s=20, alpha=0.7)
plt.plot(x_data, y_ideal, color='red', linestyle='--', label='Ideal True Function')
plt.plot(x_data, y_fit, color='green', linewidth=2, label=f'Fitted Function (m={m_fit:.2f}, c={c_fit:.2f})')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Linear Curve Fitting Demonstration')
plt.legend()
plt.grid(True)
````
